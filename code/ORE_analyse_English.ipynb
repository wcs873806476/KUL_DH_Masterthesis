```python
import pandas as pd
import pycountry
import pycountry_convert as pc

# Load the uploaded file
df = pd.read_excel("Author_Country_Distribution.xlsx")

# Country name correction dictionary
country_corrections = {
    "UK": "United Kingdom",
    "USA": "United States",
    "Russia": "Russian Federation",
    "South Korea": "Korea, Republic of",
    "North Korea": "Korea, Democratic People's Republic of",
    "Iran": "Iran, Islamic Republic of",
    "Venezuela": "Venezuela, Bolivarian Republic of",
    "Syria": "Syrian Arab Republic",
    "Vietnam": "Viet Nam",
    "Bolivia": "Bolivia, Plurinational State of",
    "Tanzania": "Tanzania, United Republic of"
}

# Function to map country to continent
def get_continent(country_name):
    try:
        name = country_corrections.get(country_name, country_name)
        country = pycountry.countries.get(name=name)
        if not country:
            country = pycountry.countries.search_fuzzy(name)[0]
        code = country.alpha_2
        continent_code = pc.country_alpha2_to_continent_code(code)
        return pc.convert_continent_code_to_continent_name(continent_code)
    except:
        return "Unknown"

# Add a new "Continent" column
df["Continent"] = df["Country"].apply(get_continent)

# Save the result to Excel
df.to_excel("Country_Continent_Statistics.xlsx", index=False)
```

```python
import pandas as pd
import networkx as nx
from itertools import combinations
import matplotlib.pyplot as plt

# Load the data
df = pd.read_csv('2authors.csv', sep=';')  # Adjust the separator according to your file format

# Drop rows with missing authors or links
df = df.dropna(subset=['author', 'link'])

# Initialize the graph
G = nx.Graph()

# Group by article link and add co-authorship edges
for link, group in df.groupby('link'):
    authors = group['author'].unique()
    if len(authors) >= 2:
        for pair in combinations(authors, 2):
            a1, a2 = sorted(pair)
            if G.has_edge(a1, a2):
                G[a1][a2]['weight'] += 1
            else:
                G.add_edge(a1, a2, weight=1)

# Print basic network statistics
print(f"Number of author nodes: {G.number_of_nodes()}")
print(f"Number of collaboration edges: {G.number_of_edges()}")

# Show top 50 most frequent co-authorship pairs
top_edges = sorted(G.edges(data=True), key=lambda x: x[2]['weight'], reverse=True)[:50]
print("\nTop 50 co-authorship pairs by collaboration count:")
for u, v, d in top_edges:
    print(f"{u} - {v}: {d['weight']} times")

# Visualize the network (subgraph of top 100 nodes by degree)
top_nodes = sorted(G.degree, key=lambda x: x[1], reverse=True)[:100]
subG = G.subgraph([n for n, _ in top_nodes])

plt.figure(figsize=(12, 12))
nx.draw_networkx(subG, with_labels=True, node_size=80, font_size=6)
plt.title("Author Collaboration Network (Top 100 Authors)")
plt.axis('off')
plt.show()
```

```python
import pandas as pd
import networkx as nx
from itertools import combinations

# Read author data
df = pd.read_csv('2authors.csv', sep=';')
df = df.dropna(subset=['author', 'link'])

# Create an undirected graph
G = nx.Graph()

# Build co-authorship edges
for link, group in df.groupby('link'):
    authors = group['author'].unique()
    if len(authors) >= 2:
        for a1, a2 in combinations(authors, 2):
            if G.has_edge(a1, a2):
                G[a1][a2]['weight'] += 1
            else:
                G.add_edge(a1, a2, weight=1)

# Export node data
nodes_data = pd.DataFrame({
    "Id": list(G.nodes()),
    "Label": list(G.nodes()),
    "Degree": [G.degree(n) for n in G.nodes()]
})
nodes_data.to_csv("gephi_nodes.csv", index=False)

# Export edge data
edges_data = pd.DataFrame([
    {"Source": u, "Target": v, "Weight": d["weight"]}
    for u, v, d in G.edges(data=True)
])
edges_data.to_csv("gephi_edges.csv", index=False)

print("Successfully exported Gephi-compatible node and edge files: gephi_nodes.csv and gephi_edges.csv")
```

```python
import pandas as pd
import networkx as nx
from itertools import combinations

# Load the data
df = pd.read_csv("2authors.csv", sep=";")
df = df.dropna(subset=["author", "link", "title"])

# === Build the full co-authorship graph ===
G = nx.Graph()
for link, group in df.groupby("link"):
    authors = group["author"].unique()
    for a1, a2 in combinations(authors, 2):
        if G.has_edge(a1, a2):
            G[a1][a2]["weight"] += 1
        else:
            G.add_edge(a1, a2, weight=1, link=link)

# === Find the connected component containing Pamela Klaassen ===
target_author = "Pamela Klaassen"
if target_author in G:
    component_nodes = nx.node_connected_component(G, target_author)
    subG = G.subgraph(component_nodes).copy()
else:
    print("Target author not found in the collaboration network.")
    subG = nx.Graph()

# === Export to Gephi files ===
nodes_df = pd.DataFrame({
    "Id": list(subG.nodes()),
    "Label": list(subG.nodes()),
    "Degree": [subG.degree(n) for n in subG.nodes()]
})
edges_df = pd.DataFrame([
    {"Source": u, "Target": v, "Weight": d["weight"]}
    for u, v, d in subG.edges(data=True)
])
nodes_df.to_csv("Pamela_network_nodes.csv", index=False)
edges_df.to_csv("Pamela_network_edges.csv", index=False)

# === Retrieve all articles involving these people (title + author) ===
people_in_subgraph = set(subG.nodes())
df_sub_articles = df[df["author"].isin(people_in_subgraph)]
article_info = df_sub_articles[["link", "title", "author"]].drop_duplicates()
article_info.to_excel("Pamela_network_articles.xlsx", index=False)

print(f"Exported {len(nodes_df)} nodes, {len(edges_df)} edges, and approximately {article_info['link'].nunique()} articles.")
```

```python
import pandas as pd

# Load data and parse version date
df = pd.read_csv("5reviewers_rounds.csv", sep=";")
df["Version Date"] = pd.to_datetime(df["Version Date"], errors="coerce")

# Remove duplicates (ensure one unique version per article)
df_unique = df.drop_duplicates(subset=["Title", "Version ID"])

# Sort by article and version date
df_sorted = df_unique.sort_values(by=["Title", "Version Date"])

# Safely extract numeric part from version ID (e.g., 'Version 1' → 1)
df_sorted["Version Number"] = (
    df_sorted["Version ID"]
    .astype(str)
    .str.extract(r"(\d+)")
    .astype(float)  # keep NaN as float
)

# Calculate time gap between adjacent versions
df_sorted["Time Gap (days)"] = df_sorted.groupby("Title")["Version Date"].diff().dt.days
df_sorted["Previous Version"] = df_sorted.groupby("Title")["Version Number"].shift()

# Keep only rows with both current and previous version numbers and time gap
df_gaps = df_sorted.dropna(subset=["Time Gap (days)", "Previous Version", "Version Number"]).copy()

# Construct labels for version intervals, e.g., "1→2"
df_gaps["Version Interval"] = (
    df_gaps["Previous Version"].astype(int).astype(str) + "→" +
    df_gaps["Version Number"].astype(int).astype(str)
)

# Average time gap between each version interval
gap_summary = df_gaps.groupby("Version Interval")["Time Gap (days)"].mean().sort_index()
print("\n📊 Average time (in days) between each version interval:")
print(gap_summary)

# Load discipline metadata
df_meta = pd.read_csv("output.csv")
df_meta["Title"] = df_meta["Title"].astype(str).str.strip()

# Merge discipline info into gap data
df_gaps = df_gaps.merge(df_meta[["Title", "Discipline"]], on="Title", how="left")

# Average gap by version interval and discipline
discipline_gap = df_gaps.groupby(["Version Interval", "Discipline"])["Time Gap (days)"].mean().unstack().T

print("\n📚 Average time between review rounds by discipline:")
print(discipline_gap)

# Count of articles for each discipline-version interval combination
discipline_counts = df_gaps.groupby(["Version Interval", "Discipline"])["Title"].count().unstack().T

print("\n📚 Number of articles per discipline per version interval:")
print(discipline_counts)
```

```python
import pandas as pd

# ================================
# 1️⃣ Load and preprocess data
# ================================
df = pd.read_csv("7funders_extracted.csv", sep=";")
df["Article ID"] = df["Link"].astype(str).str.extract(r'/articles/([^/]+)')
df["Title"] = df["Title"].astype(str).str.strip()

# ================================
# 2️⃣ Normalize funder names
# ================================
def normalize_funder_name(name):
    name = str(name).strip().lower()
    if any(k in name for k in ["horizon 2020", "h2020", "horizon2020"]):
        return "Horizon 2020"
    elif any(k in name for k in ["horizon europe", "heu", "horizoneurope"]):
        return "Horizon Europe"
    else:
        return name.title()  # Capitalize for cleanliness

df["Normalized Funder Name"] = df["Funder Name"].fillna("").apply(normalize_funder_name)

# ================================
# 3️⃣ Add Horizon project flag columns
# ================================
# Aggregate all funders per article
funders_per_article = df.groupby("Article ID")["Normalized Funder Name"].apply(lambda x: set(x.dropna()))

def classify_funding(funders):
    has_h2020 = "Horizon 2020" in funders
    has_heu = "Horizon Europe" in funders
    return pd.Series({
        "Is_H2020": has_h2020,
        "Is_HEurope": has_heu,
        "Is_Dual_Horizon": has_h2020 and has_heu,
        "Is_Unfunded": not has_h2020 and not has_heu
    })

horizon_flags = funders_per_article.apply(classify_funding)
df = df.merge(horizon_flags, left_on="Article ID", right_index=True, how="left")

# ================================
# 4️⃣ Basic stats
# ================================
print(f"📄 Total records: {len(df)}")
print(f"📄 Unique articles (by title): {df['Title'].nunique()}")

# Keep one record per article (using max funder count)
funder_per_article = df.groupby("Article ID")["Funder Count"].max()
funder_counts_unique = funder_per_article.value_counts().sort_index()

print("\n📊 Number of unique articles per funder count:")
print(funder_counts_unique)

# ================================
# 5️⃣ Top 20 most frequent funders
# ================================
top_funders = df["Normalized Funder Name"].value_counts().head(20)
print("\n🏆 Top 20 most frequent normalized funders:")
print(top_funders)

# ================================
# 6️⃣ Articles with no funding
# ================================
no_funding_df = df[df["Funder Count"] == 0][["Title"]].drop_duplicates()
print("\n📄 Articles with no funding (first 10):")
print(no_funding_df.head(10))

# ================================
# 7️⃣ Articles with both Horizon 2020 & Horizon Europe
# ================================
both_df = df[df["Is_Dual_Horizon"]][["Article ID", "Title", "Normalized Funder Name"]].drop_duplicates()
print(f"\n🌐 Number of articles funded by both Horizon 2020 & Europe: {both_df['Article ID'].nunique()}")
print(both_df.head(10))

# ================================
# 8️⃣ Articles not funded by any Horizon program + discipline
# ================================
# Load discipline info
df_meta = pd.read_csv("output.csv")
df_meta["Title"] = df_meta["Title"].astype(str).str.strip()

# Filter articles not funded by either Horizon program
neither_df = df[df["Is_Unfunded"]].copy()

# Aggregate article-level info
grouped = neither_df.groupby("Article ID").agg({
    "Title": "first",
    "Link": "first",
    "Normalized Funder Name": lambda x: "; ".join(sorted(set(x.dropna()))),
    "Funder Count": "max"
}).reset_index()

# Merge with discipline info
final_df = pd.merge(grouped, df_meta[["Title", "Discipline"]], on="Title", how="left")

# Preview
print("\n📚 Articles not funded by Horizon projects (with discipline info, first 10 rows):")
print(final_df.head(10))

# ================================
# 9️⃣ Optional: export cleaned data
# ================================
df.to_csv("7funders_cleaned.csv", index=False)
final_df.to_csv("articles_without_horizon.csv", index=False)
```
